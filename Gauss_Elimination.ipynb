```python
n = int(input("Enter number of unknowns: "))
```

```python
def sf(num):
  return float(f"{num:.{4}g}")
```

```python
A = []
print("Enter the Augmented Matrix Row wise: ")
for i in range(n):
  row = list(map(float, input().split()))
  A.append(row)
```

```python
#Forward Elimination
for k in range(n-1):
  for i in range(k+1, n):
    factor = A[i][k] / A[k][k]
    for j in range(k, n+1):
      A[i][j] = sf(A[i][j] - factor*A[k][j])
```

```python
for row in A:
  print(row)
```

```python
#Backward Substitution
x = [0]*n
x[n-1] = sf(A[n-1][n] /A[n-1][n-1])
for i in range(n-2, -1, -1):
  s = 0
  for j in range(i+1, n):
    s = s + A[i][j]*x[j]
  x[i] = sf((A[i][n] - s)/A[i][i])

for i in range(n):
  print(x)

```

```python
##euler's method
def f(x,y):
  return -2*(x**3)+12*(x**2)-20*x+8.5
```

```python
#initial step
h = float(input())
x = 0
y = 1
xf = 4
```

```python
while x < xf:
  y = y + f(x , y)*h
  x = x + h
  print(x,y)
```

```python
def lagrange(x, fx, xp):
  #given data point ar xp hocche X
   #lagrange er function. j =0,n j!=i . L = Product((x-xj)/(xi-xj))
  n = len(x) #x er size # total koyta data point oita bujhbo ei function diye
  fxp = 0 #loop e inital value zero theke start kore tai initally zero rakhsi
#outer loop e equation er code thake ar inner e thake L ber korar loop
#jehetu duita loop e dependancy ache
  for i in range(n):
    L = 1 #initially 1 dhorsi karon product one diye shuru hoy
    for j in range(n):
      if i!=j :
        factor = (xp - x[j])/(x[i]-x[j])
        L = L * factor


    term = fx[i]* L
    fxp = fxp +term
  return fxp
```

```python
n = int (input ("Order of Polynomial: ")) #order of polynomial
x = [] #x er list
fx = [] #y er list
print ("Enter x values:")
for i in range(n+1): #n+1 karon data point ekta beshi lage
  x.append(float(input())) #list er last e value insert korbe

print("Enter fx values: ")
for i in range(n+1):
  fx.append(float(input()))
```

```python
xp = float (input("Enter Interpolation point:"))
ans = lagrange( x, fx, xp)
print(ans)
```

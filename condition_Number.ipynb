```python
import numpy as np
```

```python
n = int(input("Enter Matrix size :"))
A = np.zeros((n,n)) #initialize kora hoise ,
#matrix ke initally zero rakha hoise
for i in range(n):
  for j in range(n):
    A[i][j] = float(input(f"A[{i}][{j}]="))
```

```python
print(A)
```

**DIRECT LIBRARY**

```python
if np.linalg.det(A)!=0:
  print("1-norm= ",np.linalg.cond(A,1))
  print("2-norm= ",np.linalg.cond(A,2))
  print("Infinity- norm= ",np.linalg.cond(A,np.inf))
else:
  print("Matrix is singular")
```

**MANUAL WAY**

```python
A_inv = np.linalg.inv(A)

col_sum1 = []
col_sum2 = []

for j in range(n):
  sum1 = 0
  sum2 = 0
  for i in range(n):
    sum1 = sum1 + abs(A[i][j])
    sum2 = sum2 + abs(A_inv[i][j])
  col_sum1.append(sum1)
  col_sum2.append(sum2)

condA = max(col_sum1) * max(col_sum2)
condA
```

```python
A_inv = np.linalg.inv(A)

row_sum3 = []
row_sum4 = []

for i in range(n):
  sum3 = 0
  sum4 = 0
  for j in range(n):
    sum3 = sum3 + abs(A[i][j])
    sum4 = sum4 + abs(A_inv[i][j])
  row_sum3.append(sum3)
  row_sum4.append(sum4)

condA = max(row_sum3) * max(row_sum4)
condA
```

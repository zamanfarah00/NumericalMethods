```python
#initial
xi=float(input())
n=int(input())
es=0.5*10**(2-n)
ea=100
it=1
```

```python
def f(x):
  return x**2-3*x+1
def fd(x):
  return x*2-3
```

```python
while ea >= es:
    xi1=xi-(f(xi) / fd(xi))
    ea=abs((xi1-xi)/xi1)*100
    print("Iteration no: ", it, "Root: ", xi1)
    xi = xi1
    it = it+1
```

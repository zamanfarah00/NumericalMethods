```python
def f(x,y):
  return x+y
```

```python
#initial setup
x = 0
y = 1
h = 0.05
xf = 0.1
```

```python
while x < xf:
  k1 = f(x,y)
  k2 = f(x + h/2, y + h*k1/2)
  k3 = f(x + h/2, y + h*k2/2)
  k4 = f(x + h, y + h*k3)
  y = y + h*(k1 + 2*k2 + 2*k3 + k4)/6
  x = x + h
  print(x,y)
```
